```python
import pandas as pd
dataset1=pd.read_csv('/content/insurance (1).csv')
dataset1.head()
```

```python
dataset1.info()
```

```python
from sklearn.preprocessing import LabelEncoder
lb=LabelEncoder()
dataset1['children']=lb.fit_transform(dataset1.children)
```

```python
dataset1.info()
```

```python

dataset1=dataset1.drop(columns=['sex'])
```

```python
from sklearn.preprocessing import LabelEncoder
lb=LabelEncoder()
dataset1['smoker']=lb.fit_transform(dataset1.smoker)
```

```python
from sklearn.preprocessing import LabelEncoder
lb=LabelEncoder()
dataset1['region']=lb.fit_transform(dataset1.region)
```

```python
dataset1.info()
```

```python
dataset1.describe()
```

```python
import matplotlib.pyplot as plt
plt.scatter(dataset1.children,dataset1['charges'])
plt.title("Scatter plot")
plt.xlabel("age")
plt.ylabel("charges")
plt.show()
```

```python
x=dataset1.drop(columns=['charges'])
y=dataset1['charges']
```

```python
dataset1.info()
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y, random_state=42,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(xtrain,ytrain)
lr.score(xtest,ytest)
```

```python
from sklearn.metrics import mean_squared_error as mse
predicted=lr.predict(xtest)
error=mse(ytest,predicted)**.5
print(error)
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression

models = [RandomForestRegressor(), DecisionTreeRegressor(), LinearRegression()]

errors = {}

for model in models:
    model.fit(xtrain,ytrain)
    name = type(model).__name__
    score = model.score(xtest,ytest)*100
    errors[name] = score
```

```python
errors
```

```python
import matplotlib.pyplot as plt

plt.bar(errors.keys(),errors.values())
```
